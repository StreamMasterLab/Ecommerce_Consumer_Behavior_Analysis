```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/processed/cleaned_ecommerce_data.csv"
)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Age"],
    bins=20,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Purchase_Amount"],
    bins=30,
    kde=True
)

plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Frequency_of_Purchase"],
    bins=20,
    kde=True
)

plt.title("Purchase Frequency Distribution")
plt.xlabel("Frequency of Purchase")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Customer_Satisfaction"],
    bins=10,
    kde=True
)

plt.title("Customer Satisfaction Distribution")
plt.xlabel("Customer Satisfaction")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Purchase_Category"
)

plt.title("Customers by Purchase Category")
plt.xlabel("Purchase Category")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.show()
```

```python
print(df["Purchase_Intent"].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Purchase_Intent"
)

plt.title("Purchase Intent Distribution")
plt.xlabel("Purchase Intent")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Customer_Loyalty_Program_Member"
)

plt.title("Customer Loyalty Program Membership")
plt.xlabel("Loyalty Program Member")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Purchase_Category",
    y="Purchase_Amount"
)

plt.title("Purchase Amount by Category")
plt.xlabel("Purchase Category")
plt.ylabel("Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Income_Level",
    y="Purchase_Amount"
)

plt.title("Purchase Amount by Income Level")
plt.xlabel("Income Level")
plt.ylabel("Purchase Amount")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Frequency_of_Purchase",
    y="Purchase_Amount"
)

plt.title("Purchase Frequency vs Purchase Amount")
plt.xlabel("Frequency of Purchase")
plt.ylabel("Purchase Amount")
plt.show()
```

```python
numerical_columns = [
    "Age",
    "Purchase_Amount",
    "Frequency_of_Purchase",
    "Brand_Loyalty",
    "Product_Rating",
    "Time_Spent_on_Product_Research(hours)",
    "Return_Rate",
    "Customer_Satisfaction",
    "Time_to_Decision"
]

correlation = df[numerical_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Numerical Variables")
plt.show()
```

```python
purchase_correlation = (
    correlation["Purchase_Amount"]
    .sort_values(ascending=False)
)

print(purchase_correlation)
```
