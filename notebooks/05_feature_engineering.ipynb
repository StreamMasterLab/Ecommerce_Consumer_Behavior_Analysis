```python
import pandas as pd
import numpy as np

# ============================================================
# STEP 5: FEATURE ENGINEERING
# ============================================================

# Load cleaned dataset
df = pd.read_csv(
    "../data/processed/cleaned_ecommerce_data.csv"
)

print("=" * 60)
print("FEATURE ENGINEERING")
print("=" * 60)

# ------------------------------------------------------------
# 1. Convert Time_of_Purchase to datetime
# ------------------------------------------------------------

df["Time_of_Purchase"] = pd.to_datetime(
    df["Time_of_Purchase"],
    errors="coerce"
)

# Extract useful time features
df["Purchase_Year"] = df["Time_of_Purchase"].dt.year
df["Purchase_Month"] = df["Time_of_Purchase"].dt.month
df["Purchase_Day"] = df["Time_of_Purchase"].dt.day
df["Purchase_DayOfWeek"] = df["Time_of_Purchase"].dt.dayofweek
df["Purchase_Hour"] = df["Time_of_Purchase"].dt.hour


# ------------------------------------------------------------
# 2. Create spending-related features
# ------------------------------------------------------------

df["Spending_per_Purchase_Frequency"] = (
    df["Purchase_Amount"] /
    (df["Frequency_of_Purchase"] + 1)
)

df["Engagement_Score"] = (
    df["Brand_Loyalty"] +
    df["Customer_Satisfaction"] +
    df["Product_Rating"]
) / 3


# ------------------------------------------------------------
# 3. Create research and decision behavior feature
# ------------------------------------------------------------

df["Research_to_Decision_Ratio"] = (
    df["Time_Spent_on_Product_Research(hours)"] /
    (df["Time_to_Decision"] + 1)
)


# ------------------------------------------------------------
# 4. Display new features
# ------------------------------------------------------------

new_features = [
    "Purchase_Year",
    "Purchase_Month",
    "Purchase_Day",
    "Purchase_DayOfWeek",
    "Purchase_Hour",
    "Spending_per_Purchase_Frequency",
    "Engagement_Score",
    "Research_to_Decision_Ratio"
]

print("\nNew Features Created:")
print(new_features)


# ------------------------------------------------------------
# 5. Select useful numerical features
# ------------------------------------------------------------

numerical_features = [
    "Age",
    "Purchase_Amount",
    "Frequency_of_Purchase",
    "Brand_Loyalty",
    "Product_Rating",
    "Time_Spent_on_Product_Research(hours)",
    "Return_Rate",
    "Customer_Satisfaction",
    "Time_to_Decision",
    "Purchase_Month",
    "Purchase_DayOfWeek",
    "Purchase_Hour",
    "Spending_per_Purchase_Frequency",
    "Engagement_Score",
    "Research_to_Decision_Ratio"
]

print("\nNumerical Features:")
print(numerical_features)


# ------------------------------------------------------------
# 6. Check categorical columns
# ------------------------------------------------------------

categorical_features = df.select_dtypes(
    include=["object"]
).columns.tolist()

print("\nCategorical Features:")
print(categorical_features)


# ------------------------------------------------------------
# 7. Remove Customer_ID from ML features
# ------------------------------------------------------------

# Customer_ID identifies a customer but does not represent
# meaningful behavioral information for ML.

if "Customer_ID" in df.columns:
    df = df.drop(columns=["Customer_ID"])


# ------------------------------------------------------------
# 8. Check missing values
# ------------------------------------------------------------

print("\nMissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])


# ------------------------------------------------------------
# 9. Final dataset information
# ------------------------------------------------------------

print("\nFinal Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
print(df.head())


# ------------------------------------------------------------
# 10. Save feature-engineered dataset
# ------------------------------------------------------------

output_path = "../data/processed/feature_engineered_data.csv"

df.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 60)
print("FEATURE ENGINEERING COMPLETED")
print("=" * 60)
print("Saved to:", output_path)
print("Final Shape:", df.shape)
```
