```python
import pandas as pd
import numpy as np
df = pd.read_csv(
    "../data/raw/Ecommerce_Consumer_Behavior_Analysis_Data.csv"
)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
df["Purchase_Amount"] = (
    df["Purchase_Amount"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)

print(df["Purchase_Amount"].head())
print(df["Purchase_Amount"].dtype)
```

```python
df["Time_of_Purchase"] = pd.to_datetime(
    df["Time_of_Purchase"],
    errors="coerce"
)

print(df["Time_of_Purchase"].head())
print(df["Time_of_Purchase"].dtype)
```

```python
print("Missing values before treatment:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print(df.dtypes)
```

```python
categorical_missing = [
    "Social_Media_Influence",
    "Engagement_with_Ads"
]

for col in categorical_missing:
    df[col] = df[col].fillna(df[col].mode()[0])

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)    
```

```python
numerical_columns = [
    "Age",
    "Purchase_Amount",
    "Frequency_of_Purchase",
    "Brand_Loyalty",
    "Product_Rating",
    "Time_Spent_on_Product_Research(hours)",
    "Return_Rate",
    "Customer_Satisfaction",
    "Time_to_Decision"
]

print(df[numerical_columns].describe())
```

```python
output_path = "../data/processed/cleaned_ecommerce_data.csv"

df.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset saved successfully.")
print("Location:", output_path)
print("Final shape:", df.shape)
```
