```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv(
    "../data/processed/cleaned_ecommerce_data.csv"
)

print("=" * 60)
print("CUSTOMER ANALYSIS")
print("=" * 60)

# --------------------------------------------------
# 1. Basic Customer Statistics
# --------------------------------------------------

print("\n1. BASIC CUSTOMER STATISTICS")
print("-" * 40)

print("Total Customers:", df["Customer_ID"].nunique())
print("Average Age:", round(df["Age"].mean(), 2))
print("Average Purchase Amount:", round(df["Purchase_Amount"].mean(), 2))
print("Average Purchase Frequency:", round(df["Frequency_of_Purchase"].mean(), 2))
print("Average Customer Satisfaction:", round(df["Customer_Satisfaction"].mean(), 2))


# --------------------------------------------------
# 2. Spending Analysis
# --------------------------------------------------

print("\n2. SPENDING ANALYSIS")
print("-" * 40)

print("Total Spending:", round(df["Purchase_Amount"].sum(), 2))
print("Minimum Purchase:", round(df["Purchase_Amount"].min(), 2))
print("Maximum Purchase:", round(df["Purchase_Amount"].max(), 2))

print("\nAverage Spending by Purchase Category:")
print(
    df.groupby("Purchase_Category")["Purchase_Amount"]
    .mean()
    .sort_values(ascending=False)
)


# --------------------------------------------------
# 3. Purchase Channel Analysis
# --------------------------------------------------

print("\n3. PURCHASE CHANNEL ANALYSIS")
print("-" * 40)

channel_analysis = df.groupby("Purchase_Channel").agg(
    Customers=("Customer_ID", "nunique"),
    Average_Spending=("Purchase_Amount", "mean"),
    Average_Frequency=("Frequency_of_Purchase", "mean")
).sort_values("Average_Spending", ascending=False)

print(channel_analysis)


# --------------------------------------------------
# 4. Income Level Analysis
# --------------------------------------------------

print("\n4. INCOME LEVEL ANALYSIS")
print("-" * 40)

income_analysis = df.groupby("Income_Level").agg(
    Customers=("Customer_ID", "nunique"),
    Average_Spending=("Purchase_Amount", "mean"),
    Average_Frequency=("Frequency_of_Purchase", "mean")
).sort_values("Average_Spending", ascending=False)

print(income_analysis)


# --------------------------------------------------
# 5. Loyalty Program Analysis
# --------------------------------------------------

print("\n5. LOYALTY PROGRAM ANALYSIS")
print("-" * 40)

loyalty_analysis = df.groupby(
    "Customer_Loyalty_Program_Member"
).agg(
    Customers=("Customer_ID", "nunique"),
    Average_Spending=("Purchase_Amount", "mean"),
    Average_Frequency=("Frequency_of_Purchase", "mean"),
    Average_Satisfaction=("Customer_Satisfaction", "mean")
)

print(loyalty_analysis)


# --------------------------------------------------
# 6. Purchase Intent Analysis
# --------------------------------------------------

print("\n6. PURCHASE INTENT ANALYSIS")
print("-" * 40)

intent_analysis = df.groupby("Purchase_Intent").agg(
    Customers=("Customer_ID", "nunique"),
    Average_Spending=("Purchase_Amount", "mean"),
    Average_Frequency=("Frequency_of_Purchase", "mean"),
    Average_Satisfaction=("Customer_Satisfaction", "mean")
).sort_values("Average_Spending", ascending=False)

print(intent_analysis)


# --------------------------------------------------
# 7. Customer Behavior Summary
# --------------------------------------------------

print("\n7. CUSTOMER BEHAVIOR SUMMARY")
print("-" * 40)

print("\nAverage spending by loyalty status:")
print(
    df.groupby("Customer_Loyalty_Program_Member")["Purchase_Amount"]
    .mean()
)

print("\nAverage spending by purchase intent:")
print(
    df.groupby("Purchase_Intent")["Purchase_Amount"]
    .mean()
)

print("\nAverage spending by purchase channel:")
print(
    df.groupby("Purchase_Channel")["Purchase_Amount"]
    .mean()
)


# --------------------------------------------------
# 8. Visualization — Spending by Category
# --------------------------------------------------

plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="Purchase_Category",
    y="Purchase_Amount",
    estimator="mean"
)

plt.title("Average Purchase Amount by Category")
plt.xlabel("Purchase Category")
plt.ylabel("Average Purchase Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 9. Visualization — Spending by Loyalty
# --------------------------------------------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Customer_Loyalty_Program_Member",
    y="Purchase_Amount",
    estimator="mean"
)

plt.title("Average Spending by Loyalty Program Membership")
plt.xlabel("Loyalty Program Member")
plt.ylabel("Average Purchase Amount")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# 10. Visualization — Spending by Purchase Intent
# --------------------------------------------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Purchase_Intent",
    y="Purchase_Amount",
    estimator="mean"
)

plt.title("Average Spending by Purchase Intent")
plt.xlabel("Purchase Intent")
plt.ylabel("Average Purchase Amount")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# Final
# --------------------------------------------------

print("\n" + "=" * 60)
print("CUSTOMER ANALYSIS COMPLETED")
print("=" * 60)
```
